```python
!pip install transformers torch
from transformers import pipeline

# 1. Initialize the generator (using GPT-2 as a lightweight example)
# You can replace "gpt2" with "meta-llama/Llama-3.2-1B" for more power
generator = pipeline('text-generation', model='gpt2')

# 2. Define your prompt
prompt = "The future of Artificial Intelligence in NLP is"

# 3. Generate text
# max_length: total tokens in output
# num_return_sequences: how many versions to create
results = generator(prompt, max_length=50, num_return_sequences=1, truncation=True)
#print(results)
# 4. Print the result
print(results[0]['generated_text'])
```

```python
!pip install -U diffusers transformers accelerate
import torch
from diffusers import StableDiffusionPipeline

# Load the model
model_id = "runwayml/stable-diffusion-v1-5"
pipe = StableDiffusionPipeline.from_pretrained(model_id)

# Move the pipeline to the CPU
pipe = pipe.to("cpu")
```

```python
prompt = "a futuristic city with floating neon cars, digital art style, high resolution"

# Generate image
image = pipe(prompt).images[0]

# Display image
image
```
