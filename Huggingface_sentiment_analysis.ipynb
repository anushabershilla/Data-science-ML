```python
!pip install transformers torch -q

# Load Transformer

from transformers import pipeline

classifier = pipeline("sentiment-analysis")

#Test result

result = classifier("I love learning AI")

print(result)

#multiple review

#reviews = [
 #   "This movie was amazing",
 #   "Worst movie ever",
  #  "Fantastic acting",
 #   "Terrible story",
#    "I love this movie"
#]
reviews = [
    "I love this movie",
    "Worst experience ever",
    "Amazing service",
    "Terrible product"
]
for review in reviews:
    result = classifier(review)

    print("Review:", review)
    print("Prediction:", result)
    print("-"*30)

# create dataframe

import pandas as pd

df = pd.DataFrame({
    "Review": reviews
})

df["Prediction"] = df["Review"].apply(
    lambda x: classifier(x)[0]["label"]
)

df
```
