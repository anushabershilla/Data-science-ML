```python
import pandas as pd

df = pd.DataFrame({
    'review':[
        'I love this movie',
        'This movie is amazing',
        'Fantastic acting',
        'I hate this movie',
        'Worst movie ever',
        'This movie is terrible'
    ],
    'sentiment':[1,1,1,0,0,0]
})

df

#Feature set
X = df['review']
y = df['sentiment']

#Tokenizer

from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer=Tokenizer()
tokenizer.fit_on_texts(X)
sequences=tokenizer.texts_to_sequences(X)
print(sequences)

#pading
from tensorflow.keras.preprocessing.sequence import pad_sequences
X_paded=pad_sequences(sequences,maxlen=5)
print(X_paded)

#Train test split

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_paded,y,test_size=0.2,random_state=42)

#Build NLP model
import tensorflow as tf
model=tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=50, output_dim=8),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

#compile model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

#Train model
model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=2,
    validation_split=0.2
)

#predict review
new_review = [
    "I love this film"
]

seq = tokenizer.texts_to_sequences(
    new_review
)

pad = pad_sequences(
    seq,
    maxlen=5
)

prediction = model.predict(
    pad
)

print(prediction)

#Evaluate

loss,accuracy = model.evaluate(
    X_test,
    y_test
)

print(accuracy)

prediction = model.predict(pad)

print(prediction)

print((prediction > 0.5).astype(int))
```

```python
import pandas as pd

df = pd.DataFrame({
    'review':[
        'I love this movie',
        'This movie is amazing',
        'Fantastic acting',
        'I hate this movie',
        'Worst movie ever',
        'This movie is terrible'
    ],
    'sentiment':[1,1,1,0,0,0]
})

df

#Feature set
X = df['review']
y = df['sentiment']

#Tokenizer

from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer=Tokenizer()
tokenizer.fit_on_texts(X)
sequences=tokenizer.texts_to_sequences(X)
print(sequences)

#pading
from tensorflow.keras.preprocessing.sequence import pad_sequences
X_paded=pad_sequences(sequences,maxlen=5)
print(X_paded)

#Train test split

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_paded,y,test_size=0.2,random_state=42)

#Build NLP model
import tensorflow as tf
model=tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=50, output_dim=8),
    tf.keras.layers.LSTM(16),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

#compile model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

#Train model
model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=2,
    validation_split=0.2
)

#predict review
new_review = [
    "I hate this film"
]

seq = tokenizer.texts_to_sequences(
    new_review
)

pad = pad_sequences(
    seq,
    maxlen=5
)

prediction = model.predict(
    pad
)

print(prediction)

#Evaluate

loss,accuracy = model.evaluate(
    X_test,
    y_test
)

print(accuracy)

prediction = model.predict(pad)

print(prediction)

print((prediction > 0.5).astype(int))
```
